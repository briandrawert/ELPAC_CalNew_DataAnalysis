```python
import pandas as pd
# Load enrollment data
enrollment = pd.read_csv("../1095 Data/Enrollment/CEnrollment_21-22.csv")

race_cols = [
    "StudentHispanicEthnicityIndicator",
    "StudentEthnicityMissingIndicator",
    "StudentRace1Code",
    "StudentRace2Code",
    "StudentRace3Code",
    "StudentRace4Code",
    "StudentRace5Code",
    "StudentRaceMissingIndicator",
    "RaceEthnicity"
]

# Get every unique combination and count how many students have each
combos = (
    enrollment[race_cols]
    .groupby(race_cols, dropna=False)
    .size()
    .reset_index(name="StudentCount")
    .sort_values("StudentCount", ascending=False)
)

print(f"Total unique combinations: {len(combos)}")
print(combos.to_string(index=False))
```

```python
import pandas as pd
import os

# --- Paths ---
enrollment_dir = "../1095 Data/Enrollment"

# --- Load Enrollment files ---
enrollment_files = {
    "2021-22": "CEnrollment_21-22.csv",
    "2022-23": "CEnrollment_22-23.csv",
    "2023-24": "CEnrollment_23-24.csv",
}

enrollments = {}
for label, filename in enrollment_files.items():
    enrollments[label] = pd.read_csv(os.path.join(enrollment_dir, filename), low_memory=False)
    print(f"{label}: {len(enrollments[label]):,} rows")

# --- Unique StudentSchoolCompletionStatus values across all enrollment years ---
print("\nUnique StudentSchoolCompletionStatus values:\n")
for label, df in enrollments.items():
    counts = df["StudentSchoolCompletionStatus"].dropna().astype(str).value_counts().sort_index()
    print(f"  {label}:")
    for val, count in counts.items():
        print(f"    '{val}': {count:,}")
    print()
    
```
